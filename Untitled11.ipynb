```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import time
from sklearn.linear_model import LinearRegression


@st.cache_resource
def train_ai_model():
    try:
        df = pd.read_csv('/content/Khao_sat_tien_dien_phong_tro.csv')

        house_col = df.columns[4]
        price_col = df.columns[10]
        df[house_col] = df[house_col].replace('Phòng trọ bình thường bình thường', 'Phòng trọ bình thường')

        df['fridge_num'] = df.iloc[:, 5].map({'Có': 1, 'Không': 0}).fillna(0)

        X = df.iloc[:, [1, 2, 3, 6, 7, 8]]
        X['fridge'] = df['fridge_num']
        y = df.iloc[:, 9]


        model = LinearRegression()
        model.fit(X, y)


        avg_prices = df.groupby(house_col)[price_col].mean().to_dict()

        return model, avg_prices
    except Exception as e:

        avg_prices = {"Phòng trọ bình thường": 3800, "Chung cư": 3000, "Ký túc xá": 3500, "Nhà nguyên căn": 3000}
        return None, avg_prices


model, avg_prices = train_ai_model()

st.set_page_config(page_title="AI Electricity Master", page_icon="⚡", layout="wide")

st.markdown("""
    <style>
    .stApp { background-color: #0E1117; }
    .stApp, .stMarkdown, p, li, label { color: #FFFFFF !important; }
    div[data-baseweb="select"] > div, input { color: #000000 !important; }
    [data-baseweb="popover"] *, [data-baseweb="listbox"] * { color: #000000 !important; }

    .title-text {
        background: linear-gradient(90deg, #FFD700, #FFFFFF);
        -webkit-background-clip: text;
        -webkit-text-fill-color: transparent;
        font-family: 'Segoe UI', sans-serif;
        font-weight: 800;
        text-align: center;
        text-transform: uppercase;
        letter-spacing: 3px;
        margin-bottom: 30px;
    }

    .main-card {
        background-color: #1C2128;
        padding: 30px;
        border-radius: 25px;
        border: 1px solid #30363D;
        border-top: 8px solid #FFD700;
        text-align: center;
        margin-bottom: 20px;
    }
    .price-tag { font-size: 4rem; font-weight: 900; color: #4ADE80 !important; margin: 10px 0; }

    .metric-container { display: flex; justify-content: space-between; gap: 15px; margin-top: 10px; }
    .metric-box {
        background: #161B22; padding: 15px; border-radius: 15px;
        border: 1px solid #30363D; flex: 1; text-align: center;
    }
    .metric-label { color: #8B949E !important; font-size: 0.8rem; text-transform: uppercase; }
    .metric-value { color: #FFD700 !important; font-size: 1.4rem; font-weight: bold; }

    .ai-advice-title { color: #FFD700 !important; font-weight: bold; font-size: 1.2rem; margin-top: 25px; margin-bottom: 15px; display: flex; align-items: center; gap: 10px; }
    .advice-card { background-color: #161B22; border: 1px solid #30363D; border-radius: 12px; padding: 15px; margin-bottom: 10px; display: flex; align-items: center; gap: 15px; }
    .advice-icon { font-size: 1.5rem; }
    .advice-text { color: #FFFFFF !important; font-size: 0.95rem; margin: 0; line-height: 1.4; }

    div.stButton > button:first-child {
        background: linear-gradient(135deg, #FFD700 0%, #F0B232 100%);
        color: #000000 !important;
        border-radius: 12px; width: 100%; font-weight: bold; border: none;
    }
    .stNumberInput label, .stSelectbox label, .stSlider label, .stRadio label { color: #FFD700 !important; font-weight: 600 !important; }
    </style>
    """, unsafe_allow_html=True)

st.markdown('<h1 class="title-text">⚡ HỆ THỐNG DỰ TOÁN TIỀN ĐIỆN AI</h1>', unsafe_allow_html=True)

col_input, col_result = st.columns([1, 1.2], gap="large")

with col_input:
    st.markdown("### 📋 THÔNG TIN NHẬP LIỆU")
    c1, c2 = st.columns(2)
    with c1:
        num_people = st.number_input("👥 SỐ NGƯỜI Ở", 1, 15, 2)
        dien_tich = st.number_input("📏 DIỆN TÍCH (M2)", 10, 150, 25)
    with c2:
        house_type = st.selectbox("🏠 LOẠI HÌNH NƠI Ở", options=list(avg_prices.keys()))
        lau = st.number_input("🏢 TẦNG LẦU", 0, 10, 0)

    st.markdown("---")
    st.markdown("### ⚙️ CẤU HÌNH THIẾT BỊ")
    num_ac = st.slider("Số lượng máy lạnh", 0, 5, 1)
    ac_hours = st.slider("⏱️ Thời gian sử dụng máy lạnh (giờ/ngày)", 0, 24, 8)

    c3, c4 = st.columns(2)
    with c3:
        num_fans = st.number_input("🌬️ SỐ LƯỢNG QUẠT", 0, 10, 2)
    with c4:
        has_fridge_text = st.radio("🧊 CÓ TỦ LẠNH KHÔNG?", ["Có", "Không"], horizontal=True)

    has_fridge = 1 if has_fridge_text == "Có" else 0
    predict_btn = st.button("🚀 XÁC NHẬN DỰ TOÁN BẰNG AI")

with col_result:
    if predict_btn:
        with st.spinner('Mô hình Hồi quy tuyến tính đang tính toán...'):
            time.sleep(0.8)

            if model is not None:
                input_data = np.array([[num_people, dien_tich, lau, num_fans, num_ac, ac_hours, has_fridge]])
                total_money = model.predict(input_data)[0]
                total_money = max(0, total_money)
            else:
                total_money = (num_people * 50 + num_ac * 100) * avg_prices.get(house_type, 3500)

            unit_price = avg_prices.get(house_type, 3500)
            total_kwh = total_money / unit_price if unit_price > 0 else 0

            st.markdown(f"""
                <div class="main-card">
                    <p style="color: #FFD700; font-weight: bold;">DỰ DỰ TOÁN BỞI THUẬT TOÁN HỒI QUY</p>
                    <h1 class="price-tag">{total_money:,.0f} <span style="font-size: 1.5rem">VNĐ</span></h1>
                    <p style="color: #8B949E; font-size: 0.9rem;">Kết quả phân tích từ dữ liệu thực tế khảo sát</p>
                </div>
                """, unsafe_allow_html=True)

            st.markdown(f"""
                <div class="metric-container">
                    <div class="metric-box">
                        <div class="metric-label">⚡ Tiêu thụ ~</div>
                        <div class="metric-value">{total_kwh:.1f} kWh</div>
                    </div>
                    <div class="metric-box">
                        <div class="metric-label">❄️ Máy lạnh</div>
                        <div class="metric-value">{ac_hours}h/ngày</div>
                    </div>
                    <div class="metric-box">
                        <div class="metric-label">📊 Độ tin cậy</div>
                        <div class="metric-value">Cao</div>
                    </div>
                </div>
                """, unsafe_allow_html=True)

            st.markdown('<div class="ai-advice-title">💡 LỜI KHUYÊN TỪ AI</div>', unsafe_allow_html=True)
            advices = []
            if ac_hours > 8: advices.append("Máy lạnh đang tiêu tốn nhiều nhất. Hãy cài đặt nhiệt độ từ 25-27°C để tiết kiệm điện hiệu quả.")
            if total_money > 1500000: advices.append("Chi phí của bạn đang cao hơn mức trung bình. Hãy cân nhắc thay thế thiết bị cũ bằng Inverter.")
            if has_fridge: advices.append("Hạn chế mở tủ lạnh quá lâu và không để thực phẩm còn nóng vào tủ.")

            if not advices: advices.append("Mức tiêu thụ điện của bạn đang rất tốt. Hãy tiếp tục duy trì!")

            for advice in advices:
                st.markdown(f"""<div class="advice-card"><div class="advice-icon">💡</div><p class="advice-text">{advice}</p></div>""", unsafe_allow_html=True)
    else:
        st.info("Nhấn nút để AI dự toán dựa trên thuật toán Hồi quy tuyến tính.")
```
